```python
import os
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
data_folder = "a1_data"
processed = {}
```

```python
for filename in os.listdir(data_folder):
    if filename.endswith(".txt"):
        file_path = os.path.join(data_folder, filename)
        
        with open(file_path, "r", encoding="utf-8") as f:
            text = f.read()
        
        tokens = word_tokenize(text)
        clean = [
            stemmer.stem(word.lower()) 
            for word in tokens 
            if word.isalpha() and word.lower() not in stop_words
        ]

        processed[filename] = clean

# Preview one file
print("Sunwoo Lee - Z23597551")
for filename, tokens in processed.items():
    print(f"\n--- {filename} ---")
    print(tokens[:20])
    print("Total tokens:", len(tokens))
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
doc_names = list(processed.keys())
docs_as_text = [" ".join(processed[name]) for name in doc_names]

vectorizer = TfidfVectorizer(
    tokenizer=str.split,      
    preprocessor=None,
    lowercase=False,
    token_pattern=None    
)

tfidf_matrix = vectorizer.fit_transform(docs_as_text)  # shape: (num_docs, num_terms)
terms = vectorizer.get_feature_names_out()

doc_word_tfidf = pd.DataFrame(tfidf_matrix.toarray(), index=doc_names, columns=terms)
tall_tfidf = doc_word_tfidf.T

tall_tfidf.to_csv("doc_word_tfidf.csv")

print("Sunwoo Lee - Z23597551")
print("Matrix shape:", tall_tfidf.shape)
tall_tfidf.iloc[:, :30]   
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
sim = cosine_similarity(tfidf_matrix)
doc_sim = pd.DataFrame(sim, index=doc_names, columns=doc_names)
print("Sunwoo Lee - Z23597551")
doc_sim.iloc[:10, :10]
```
